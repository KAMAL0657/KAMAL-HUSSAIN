# UAV Maize Crop Image Preprocessing Simulation

This notebook simulates the image preprocessing pipeline for UAV maize crop images.

**Steps covered:**
1. Create a dummy high-resolution UAV image
2. Split it into 224x224 sub-images
3. Simulate blur detection (20%)
4. Print statistics for ML training

```python

import random

# Step 1: Dummy high-resolution UAV image size
image_width = 6000
image_height = 3376

# Step 2: Sub-image size
patch_size = 224

# Calculate number of patches
patches_x = image_width // patch_size
patches_y = image_height // patch_size

total_patches = patches_x * patches_y

# Step 3: Simulate blur detection (20%)
blurry_percentage = 0.20
blurry_images = int(total_patches * blurry_percentage)
clear_images = total_patches - blurry_images

# Random blur labeling (simulation)
labels = ["Blurry"] * blurry_images + ["Clear"] * clear_images
random.shuffle(labels)

# Step 4: Print results
print("UAV Maize Crop Image Preprocessing Simulation")
print("-" * 50)
print(f"Original Image Size       : {image_width} x {image_height}")
print(f"Sub-image Size            : {patch_size} x {patch_size}")
print(f"Patches along width       : {patches_x}")
print(f"Patches along height      : {patches_y}")
print(f"Total sub-images created  : {total_patches}")
print(f"Blurry images detected    : {blurry_images}")
print(f"Clear images for training : {clear_images}")
```
