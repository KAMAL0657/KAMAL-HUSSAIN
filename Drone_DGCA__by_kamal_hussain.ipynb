# Drone Classification using DGCA Categories



## DGCA Drone Categories
- **Nano**: up to 250 g
- **Micro**: 250 g – 2000 g
- **Small**: 2000 g – 25000 g
- **Medium**: 25000 g – 150000 g
- **Large**: more than 150000 g

Medium and Large drones are generally used for **agricultural spraying**.

```python
# Input drone weight in grams
weight = 5000.0  # Sample weight for demonstration

# Drone classification
if weight <= 250:
    category = 'Nano'
    certification = 'No certification required'
    spraying = 'Not suitable for agricultural spraying'

elif weight <= 2000:
    category = 'Micro'
    certification = 'Basic DGCA registration'
    spraying = 'Not suitable for agricultural spraying'

elif weight <= 25000:
    category = 'Small'
    certification = 'Remote Pilot Certificate (RPC) required'
    spraying = 'Not suitable for agricultural spraying'

elif weight <= 150000:
    category = 'Medium'
    certification = 'Advanced RPC + DGCA approval required'
    spraying = 'Suitable for agricultural spraying'

else:
    category = 'Large'
    certification = 'Special DGCA permission required'
    spraying = 'Suitable for agricultural spraying'

# Output result
print('\nDrone Category:', category)
print('Certification Required:', certification)
print('Agricultural Spraying:', spraying)
```

## Conclusion
This program helps in identifying the drone category based on weight, the required DGCA certification, and whether it can be used for agricultural spraying. The logic is simple and suitable for **students and beginners**.
